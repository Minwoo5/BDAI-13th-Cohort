## 과제 2. 미니 문서 진단하기 (필수) 🔎

이번에는 문장이 아니라 **여러 줄짜리 미니 문서**입니다. 청소하기 전에, 실습에서 배운 **관찰 도구**로 진단해 봅시다.

> 따옴표 세 개(`"""`)로 감싸면 여러 줄 문자열이 됩니다 — 다음 주에 제대로 배우니, 오늘은 그대로 실행만 하세요.

```python
# 1️⃣ 미니 문서를 담습니다 (그대로 실행)
mini_doc = """주문 취소는 배송 준비 전까지 가능합니다.
배송 준비가 시작되면 취소 대신 반품을 이용해 주세요.
반품 신청은 마이페이지 또는 FAQ 게시판에서 할 수 있습니다.
교환 상품의 배송비는 무료입니다.
자세한 내용은 FAQ 게시판을 확인해 주세요."""

print(mini_doc)
```

```python
# 2️⃣ 진단 5종 — 빈칸을 채워 미니 문서를 진단해 보세요

# ① 글자 수 재기
print(f"글자 수: {len(____)}자")

# ② "반품"이라는 말이 들어 있나요? (빈칸에 in)
print("반품이 들어있나요?:", "반품" ____ mini_doc)

# ③ "배송"은 몇 번 나오나요? (빈칸에 도구 이름)
print(f"'배송' 등장 횟수: {mini_doc.____('배송')}회")

# ④ 앞에서 15글자만 미리보기 (빈칸에 숫자)
print("미리보기:", mini_doc[:____])

# ⑤ 소문자 faq 로 검색하면? → 실패! 대소문자를 통일하고 다시 검색해 보세요 (빈칸에 도구 이름)
print("faq 검색(통일 전):", "faq" in mini_doc)
print("faq 검색(통일 후):", "faq" in mini_doc.____())
```

💡 힌트
- ① 미니 문서가 담긴 상자 이름은? / ② 포함 확인 도구는 단어 사이에 끼워 씁니다
- ③ 횟수를 세는 도구는 `count` / ④ 앞에서 n글자는 `[:n]` / ⑤ 소문자 통일 도구는 `lower`

**예상 결과:** ②는 `True`, ③은 `3회`, ⑤는 통일 전 `False` → 통일 후 `True` 가 나와야 합니다.
